```python
import pandas as pd

archivo = pd.read_csv('medallas.csv')
df = pd.DataFrame(archivo)
print(df.head())
```

```python
#datos_limpios = df.dropna()
#Tambien con el metodo fillna() usando el parametro inplace=True para modificar el dataframe original
df.fillna(0, inplace=True)
#print(datos_limpios)
```

```python
df['Oro'] = df['Oro'].astype(int)
df['Plata'] = df['Plata'].astype(int)
df['Bronce'] = df['Bronce'].astype(int)
print(df.head())
```

```python
df_ordenado = df.sort_values(by='Oro', ascending=False) #por defecto es de menor a mayor, con ascending=False es al reves
print(df_ordenado.head(3))
```

```python
paises_con_mas_de_diez_medallas = df[df['Total'] > 10]
print(paises_con_mas_de_diez_medallas)
```

```python
filtro = df['Total'] > 10
mas_de_10_medallas = df[filtro]
mas_de_10_medallas_ordenado = mas_de_10_medallas.sort_values(by='Total', ascending=False)
print(mas_de_10_medallas_ordenado)
```

```python
desviacion_estandar_total = df['Total'].std()
print(f"Desviación estándar de la columna 'Total': {desviacion_estandar_total}")
```
