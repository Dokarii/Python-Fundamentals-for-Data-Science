```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'Nombre': ['Juan', 'Ana', 'Pedro'], 'Edad': [25, 30, 35]})
df1
```

```python
df2 = pd.DataFrame({'Nombre': ['Luis', 'Maria'], 'Edad': [28, 32]})
```

```python
df3 = pd.concat([df1, df2], axis=0, ignore_index=True)
df3
```

```python

datos_cliente = pd.DataFrame({'Nombre': ["Ana", "Luis", "Marta"], 'Edad': [34, 45, 28]})
compras_cliente = pd.DataFrame({'Producto': ["Libro", "Lápiz", "Cuaderno"], 'Precio': [15.50, 0.50, 2.00]})

info_cliente = pd.concat([datos_cliente, compras_cliente], keys=['Datos del Cliente', 'Compras del Cliente'], axis=1)
info_cliente
```
