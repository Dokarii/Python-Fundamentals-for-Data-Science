```python
import pandas as pd
```

```python
df = pd.read_csv("Top-Películas.csv")
df.info()
```

```python
df_ordenado = df.sort_values(by='rating', ascending=False) #por defecto es de menor a mayor, con ascending=False es al reves
print(df_ordenado.head(3))
```

```python
df_agrupado = df.groupby('género') ['recaudación(M)'].mean()
print(df_agrupado)
```

```python
datos = {
    'titulo': ['Pelicula A', 'Pelicula B', 'Pelicula C', 'Pelicula D'],
    'año': [2001, 2000, 2005, 2010],
    'rating': [7.2, 6.5, 8.1, 7.5]
}

df = pd.DataFrame(datos)

df_ordenado = df.sort_values(by=['rating', 'año'], ascending=[False, True])
print(df_ordenado)
```
